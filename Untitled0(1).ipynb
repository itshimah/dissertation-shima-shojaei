```python
# =============================================================================
# THESIS: AI-Powered E-commerce Personalization through Big Data Mining
# AUTHOR: Shima
# DESCRIPTION: Advanced NLP Pipeline for Sentiment-Driven Recommendations
# DATA: 1,000,000 Amazon Reviews (Big Data Scale)
# =============================================================================

import pandas as pd
import numpy as np
import joblib
import gc
import logging
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_curve, auc
from google.colab import drive

# 1. INITIALIZATION & DRIVE MOUNT
# ---------------------------------------------------------
drive.mount('/content/drive')
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(message)s', datefmt='%H:%M:%S')
logger = logging.getLogger()

DATA_PATH = '/content/drive/MyDrive/Dissertation/train.csv'
MODELS_DIR = '/content/drive/MyDrive/Dissertation/'

def run_final_dissertation_pipeline():
    # PHASE 1: BIG DATA INGESTION (Exactly 1 Million Rows)
    # ---------------------------------------------------------
    logger.info("⏳ Phase 1: Mining 1,000,000 records from Big Data source...")
    try:
        df = pd.read_csv(DATA_PATH, header=None, nrows=1000000, names=['Label', 'Title', 'Text'])
        df['Full_Text'] = df['Title'].fillna('') + " " + df['Text'].fillna('')
        df['Label'] = df['Label'].map({1: 0, 2: 1}).astype('int8')
        df.drop(['Title', 'Text'], axis=1, inplace=True)
        gc.collect()
        logger.info(f"✅ Data Ingestion Complete. Total Records: {len(df)}")
    except Exception as e:
        logger.error(f"❌ Critical Error during data loading: {e}")
        return

    # PHASE 2: DATA SPLITTING
    # ---------------------------------------------------------
    logger.info("⏳ Phase 2: Splitting Data into 800k (Train) and 200k (Test)...")
    X_train, X_test, y_train, y_test = train_test_split(
        df['Full_Text'], df['Label'], test_size=0.2, random_state=42, stratify=df['Label']
    )
    del df
    gc.collect()

    # PHASE 3: FEATURE EXTRACTION (N-Grams for Personalization)
    # ---------------------------------------------------------
    logger.info("⏳ Phase 3: Extracting features (Unigrams & Bigrams)...")
    tfidf = TfidfVectorizer(max_features=1500, stop_words='english', ngram_range=(1, 2))
    X_train_vec = tfidf.fit_transform(X_train)
    logger.info("✅ Feature Mining Complete.")

    # PHASE 4: COMPARATIVE MODEL TRAINING (Mining Insights)
    # ---------------------------------------------------------
    # Logistic Regression (Baseline - Usually very strong for text)
    logger.info("⏳ Phase 4.1: Training Logistic Regression Model...")
    lr_model = LogisticRegression(max_iter=1000, n_jobs=-1, random_state=42)
    lr_model.fit(X_train_vec, y_train)

    # Random Forest (Primary Model for Pattern Recognition)
    logger.info("⏳ Phase 4.2: Training Random Forest (100 Trees)...")
    rf_model = RandomForestClassifier(
        n_estimators=100,
        max_depth=25,
        n_jobs=-1,
        random_state=42,
        verbose=1 # This will show the Parallel logs you wanted
    )
    rf_model.fit(X_train_vec, y_train)

    # PHASE 5: SAVING THE PERSONALIZATION PIPELINE (For Web App)
    # ---------------------------------------------------------
    # We save as a 'Pipeline' so the App only needs one file to work
    final_pipeline = Pipeline([
        ('tfidf', tfidf),
        ('classifier', lr_model) # Using the higher-accuracy model for the App
    ])
    joblib.dump(final_pipeline, MODELS_DIR + 'ecommerce_personalization_model.pkl')
    logger.info("💾 AI Personalization Pipeline saved for Web Application.")

    # PHASE 6: EVALUATION & TEXTUAL REPORTS
    # ---------------------------------------------------------
    X_test_vec = tfidf.transform(X_test)

    y_pred_rf = rf_model.predict(X_test_vec)
    y_pred_lr = lr_model.predict(X_test_vec)

    acc_rf = accuracy_score(y_test, y_pred_rf)
    acc_lr = accuracy_score(y_test, y_pred_lr)

    # Display same output as requested
    print("\n" + "="*60)
    print(f"🏆 PRIMARY MODEL (Random Forest) ACCURACY: {acc_rf * 100:.2f}%")
    print(f"📊 BASELINE MODEL (Logistic Regression) ACCURACY: {acc_lr * 100:.2f}%")
    print("="*60 + "\n")

    print("--- Detailed Classification Report (Best Model) ---\n")
    print(classification_report(y_test, y_pred_lr))

    # PHASE 7: VISUALIZATIONS FOR DISSERTATION
    # ---------------------------------------------------------
    generate_visuals(y_test, y_pred_lr, rf_model, tfidf)

def generate_visuals(y_test, y_pred, rf_model, tfidf):
    # 1. Confusion Matrix
    plt.figure(figsize=(6,5))
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Greens')
    plt.title('Sentiment Prediction (Confusion Matrix)')
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.show()

    # 2. Top 20 Feature Importance (Mining Insights)
    # Fixed the 'palette' warning by assigning 'hue'
    feature_names = tfidf.get_feature_names_out()
    importances = rf_model.feature_importances_
    feat_df = pd.DataFrame({'Feature': feature_names, 'Importance': importances})
    top_20 = feat_df.sort_values(by='Importance', ascending=False).head(20)

    plt.figure(figsize=(10,6))
    sns.barplot(x='Importance', y='Feature', data=top_20, hue='Feature', palette='viridis', legend=False)
    plt.title('Top 20 Mining Features for Personalization')
    plt.show()

if __name__ == "__main__":
    run_final_dissertation_pipeline()
```
